```python
01) 버블 정렬 
- 인접한 두요소를 비교하여 정렬하는 알고리즘
- 두 요소의 크기를 비교하여 필요하면 서로의 위치를 교환
- 한번의 반복이 끝나면 가장 큰값이 뒤쪽으로 이동
```

```python
초기값: [65,34,25,12]
1단계
[34,65,25,12]
[34,25,65,12]
[34,25,12,65]
2단계
[25,34,12,65]
[25,12,34,65]
3단계
[12,25,34,65]
```

```python
01. 장점
- 구현 방법이 단순하고 이해하기 쉬움
- 데이터가 적은 경우 사용하기 적합

02. 단점
- 데이터가 많아지면 정렬속도가 느림
- 인접한 요소를 반복적으로 비교하고 교환해야함
-평균 및 최악의 시간복잡도: O(n^2)
```

```python
def bubble_sort(array):#[65,34,25,12]
    n = len(array) #4
    for i in range(n): #n=4 -> i=0,1,2,3
        print(f"{i}단계: {array}")
        for j in range(0, n-i-1): 
            #0단계 range(0, 3) =>j: 0,1,2
            #1단계: range(0,2) => j: 0,1 
            #n-i-1: 이미 정렬된 위치의 값을 비교하지 않도록 범위를 줄임!
            if array[j] > array[j+1]:
                array[j], array[j+1] = array[j+1], array[j]
    return array
array = [65,34,25,12]
print("sorted:", bubble_sort(array))
```

```python
02) 퀵정렬
- 피벗(pivot)을 기준으로 데이터를 두그룹으로 나누어 정렬하는 알고리즘
- 피벗보다 작은값 -> 왼쪽 그룹
- 피벗보다 큰값 -> 오른족 그룹
- 나누어진 각 그룹에 대해 같은 과정을 재귀적으로 반복

01.장점
일반적으로 정렬속도가 빠름
평균 시간복잡도: O(n log n)
- 큰데이러를 정렬하는데 효율적
02. 단점
- 피벗 선택이 좋지 않으면 성능이 크게 떨어질 수 있음
- 최악의 경우 시간 복잡도: O(n^2)
- 재귀 호출을 사용하므로 재귀 깊이가 깊어질 수 있음
03. 퀵정렬의 동작  원리 
[1]  피벗선택
- 피벗을 첫번째, 마지막, 중간값 중 어떤 값이든 선택 가능
-> 정렬할 데이터에서 기준으로 사용함

[2]분할
- 피벗을 기줏으로 리스트를 두부분으로 나눔
- 피벗보다 작은 요소들은 피벗의 왼쪽부분에, 피벗보다 큰 요소들은 피벗의 오른쪽 부분에 위치
```

```python
def qsort(arr):# arr = [24,26,2,16,32,31,25]
    if len(arr) < 2: 
        return arr
    pivot = arr[0]#24
    low = []
    high = []
    for value in arr[1:]:#[26,2,16,32,31,25]
        if value < pivot:
            low.append(value) #[2,16]
        else:
            high.append(value) #[26,32,31,25]
    #출력
    print(f"{low} + [{pivot}] +{high} (pivot:{pivot})")
    return  qsort(low) + [pivot] + qsort(high)
                #[2,16] + [24] +  [25,26,31,32]
my_list = [24,26,2,16,32,31,25]
print("정렬전")
print(my_list)
print('정렬과정')
sorted_list = qsort(my_list)
print("정렬 결과")
print(sorted_list)
```

```python
- 데이터에서 원하는 값을 찾아내는 과정
- 많은 데이터 중에서 특정값이나 조건에 맞는 데이터를 찾는것
01) 선형검색(Linear Search)
- 순차검색(Sequential Search)이라고도 함
- 데이터가 정렬되어 있지 않아도 검색가능
- 데이터의 첫번째 요소부터 순서대로 하나씩 비교하는 방식
- 찾는 값과 일치하면 검색을 종료 
- 원하는 값을 찾을 때까지 데이터를 차례대로 확인
시간복잡도: O(n)
01. 장점
- 구현이 간단하고 쉬움
- 데이터가 작은 경우 효율적
- 찾고자 하는 값이 앞부분에 있는 경우 빠르게 검색 가능
02. 단점
- 데이터의 크기가 커질수록 검색 시간이 선형적으로 증가
- 데이터가 많은경우 비효율적
03. 선형검색의 종료 조건
- 검색할 값과 같은 원소를 찾은 경우->검색 성공
- 배열의 끝까지 확인했지만 값을 찾지 못한 경우 -> 검색 실패
```

```python
def sequential_s(arr, target):#[10,32,25,47,53,68,81,72]
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1#"없음"
array = [10,32,25,47,53,68,81,72]
print(sequential_s(array,72))
print(sequential_s(array,50))
```

```python
02) 이진검색(Binary search)
- 정렬된 배열을 반복적으로 절반씩 나누어 원하는 값을 찾는 알고리즘
- 데이터가 반드시 정렬되어 있어야함
- 중간값(mid)과 target을 비교하여 탐색 범위를 절반으로 줄임
- target이 mid값보다 작으면 왼쪽 영역 탐색
- target이 mid값보다 크면 오른쪽 영역 탐색
- 시간복잡도: O(log n)
```

```python
01. 이진탐색 방법
- 자료를 오름차순 정렬
- 자료의 중간값(mid)과 찾고자하는 값(target)을 비교
- target과 mid의 대소관계에 따라 탐색 범위를 절반으로 줄임
- target과 mid가 같아질 때까지 1,2을 반복
1. target이 mid보다 작은 경우 
 - 오른쪽 영역을 탐색 범위에서 제거
 - right = mid -1 
2. target이 mid보다 큰경우
 - 왼쪽 영영을 탐색 범위에서 제거
 - left = mid +1
02. 장점
- 탐색 범위를 매번 절반으로 줄이기 때문에 검색 속도가 빠름
- 데이터가 많을 수록 선형검색보다 효율적
03. 단점
- 데이터가 반드시 정렬되어 있어야함
- 데이터의 삽입이나 삭제가 빈번한 경우 정렬상태를 유지하기 어려움으로 사용시 번거로움
```

```python
04. 이진탐색 구현하기 
```

```python
def binary_s(arr, target):#arr:[10,25,28,47,53,68,72,81]
    left = 0
    right =len(arr) -1 #8-1 ->7
    while left <= right:
        mid = (left + right) //2
        #1차: 0+7 // 2 -> 3
        #2차: 4+7 // 2 => 5
        #3차: 6+7 // 2 =>6
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid +1
        else: 
            right = mid -1
    return -1
array = [10,25,28,47,53,68,72,81]
print(binary_s(array, 72))
```
