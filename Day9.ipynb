## 01 ㅣ 생성자(__init__)

```python
class Animal:  
    def __init__(self, first, second):  # __init__에서 초기값 세팅 => 생성자: 발 개수, 꼬리 개수 
        self.first = first   # 발 개수 속성
        self.second = second  # 꼬리 개수 속성 
    def say(self):  # 말하기 메서드
        return f"나는 발이 {self.first}개이고, 꼬리는 {self.second}개야."
cat = Animal(4,1) # first=4, second=1 자동 저장 # cat = 주소/ 객체를 생성하자 마자 init()이 작동하므로 인수 2개 전달 해야함
print(cat.first)
print(cat.say()) # 반환된 문장 출력 , # say는 기능(메서드)이므로 괄호를 열고 닫아서 사용 
```

```python
class Animal:
    def __init__(self, first, second, say):
        self.first = first
        self.second = second
        self.say = say # 속성 이름이 say
    def say1(self): # 말하기 메서드
        return f"나는 발이 {self.first}개이고, 꼬리는 {self.second}개야."
cat = Animal(4,1, "나는 말도해")  # say 속성에 인수 전달 (저장 )
print(cat.first)
print(cat.say) # () 안붙이면 cat 의 속성 say에 저장된 값을 본다 # 속성 → 저장된 값
print(cat.say1()) # () 붙이면say1라는메서드(기능)을 사용함 # 메서드 → 괄호로 실행
```

```python
- cat 자체를 출력하면 주소값
  => cat.first는 "cat의 주소에 가서 그 안의 first에 뭐가 들어있나 봐줘"라는 뜻

- 속성과 메서드는 접근법이 완전히 다름: 
  cat.say → "저기 있는 속성값을 볼 거야"
  cat.say1() → "그 객체가 가지고 있는 메서드를 쓸 거야"
```

```python
01) 상속 사용하기
[작성법]
class 부모클래스명:
    사용할 메서드1               # 부모의 기능 1
    사용할 메서드2               # 부모의 기능 2
class 자식클래스명(부모클래스명): # 자식 클래스는 메서드 1,2,3,4를 모두 사용 가능
    사용할 메서드3               # 자식이 추가한 기능 3
    사용할 메서드4               # 자식이 추가한 기능 4
 
```

```python
class Person:   
    def __init__(self, name, age): # 부모의 생성자
        self.name = name 
        self.age = age  
    def greet(self):     # 부모의 메서드
        print(f'Hello, My name is {self.name}')
class Student(Person): # Person을 상속받는 자식 클래스
    def study(self):   # 자식이 추가한 메서드
        print(f'{self.name}학생은 공부를 합니다.') # 부모 생성자가 저장한 self.name 사용

s = Student('Alice', 20) # 부모 클래스를 모두 물려받기 떄문에 init()동작 -> 인수 필수로 넣어야함
s.greet()    # 부모에게 물려받은 메서드
s.study()    # 자식의 메서드
```

```python
class Calculator: # 부모가 될 계산기
    def __init__(self, first, second): # 생성자
        self.first = first
        self.second = second
    def add(self):
        return self.first + self.second
    def sub(self):
        return self.first - self.second
    def mul(self):
        return self.first * self.second
    def div(self):
        return self.first / self.second    
```

```python
class FiveCal(Calculator): # 자식 클래스 -> 위의 클래스 상속
    def pow(self):     # 거듭제곱 메서드 추가 # Calculator의 기능 4개 + 새 기능
        return self.first ** self.second

a = FiveCal(4,2) # 상속받기에 init()동작 -> 두가지 값을 정확히 전달해야함
print(a.add())   # 물려받은 메서드
print(a.pow())   # 추가한 메서드
```

```python
class Animal:        # 부모 1
    def bark(self):
        print('bark')
class Mammal:        # 부모 2
    def feed(self):
        print('feed')
class Cat(Animal, Mammal): # 다중 상속 => Cat 클래스가 위의 두가지 클래스 모두 물려받음
    def name(self):
        print('Cat')
cat1 = Cat()
cat1.bark() # Animal에게 받은 기능
cat1.feed() # Mammal에게 받은 기능
cat1.name() # 자기 기능 — ⭐ 메서드는 괄호 필수
```

```python
class Person:           # 부모 1
    def greeting(self):
        print('안녕하세요')
class Worker:           # 부모 2
    def greeting2(self):
        print('저는 일하는 중입니다.')
class Student:          # 부모 3
    def study(self):
        print('저는 공부하는 중입니다.')
class Intern(Person, Worker, Student): # 부모 3개를 상속
    def greeting3(self):
        super().greeting()  # 부모 중 greeting을 가진 Person의 메서드 호출 
        # super() -> 위의 class들의 메서드 중에 greeting기능을 사용하겠다.
        print('저는 인턴입니다.') 
jane = Intern()
jane.greeting3() 
```

```python
⭐ super():  상위 클래스의 메서드를 가져다 쓸 때 사용
=> super()는 상위 클래스를 가리키는 것. 
('지금 상위 클래스 중에 greeting이라는 게 있어, 그러면 그걸 가져다가 쓸 거야' 명시)
```

## 04 ㅣ 오버라이딩이란? 

```python
04 ㅣ 오버라이딩이란? 
- 무시하다, 우선하다 (override)
-> 부모클래스의 메서드를 무시하고 새로운 메서드를 만든다는 것
-> 부모클래스의 메서드를 자식클래스에서 재정의
```

```python
# 자식 이기는 부모없다 → 같은 이름이면 자식의 메서드가 우선한다는 비유
```

```python
class Parent:       # 부모
    def call(self): # 부모의 call
        print('부모호출')
class Son(Parent):  # 자식
    def call(self): # 같은 이름으로 다시 정의 → 오버라이딩 / call 기능은 가져다가 쓰나 일부분을 다시 재정의 
        print('자식호출')

a = Son() # Son 이라는 클래스 사용하여 객체 생성
a.call()  # 자식의 call이 우선 → 자식호출
```

```python
class Parent:
    def say(self):
        pass
    def sound(self):
        pass
    def yes(self):
        pass    
    def call(self):
        print('부모호출') 
class Son(Parent):
    def call(self): 
        print('자식호출') # 다른 부분은 물려받고 특정 부분만 코드 수정
    def call2(self):
        print('')

a =Son() 
a.call()
```

```python
## 오버라이딩을 사용하는 이유
- 자식클래스에 맞게 부모의 메서드를 재정의하기 위해 
- 공통기능은 재사용하고 필요한 부분만 수정하기 위해
=> 객체마다 다른 동작을 수행하게 하기 위해서
```

```python
class Animal: # 부모
    def sound(self): # 공통 기능
        print('오버라이딩 전') # 부모 버전
        print('멍멍')
class Dog(Animal):   # Animal 상속
    def sound(self): # 오버라이딩(재정의)
        print('오버라이딩 후') 
        print('멍멍멍')
class Cat(Animal):
    def sound(self): # 오버라이딩
        print('오버라이딩 후')
        print('야옹')

for x in [Animal(), Dog(), Cat()]: # 객체 3개를 리스트로 돌며
    x.sound()        # 같은 호출이 객체마다 다른 동작 (오버라이딩의 목적)
```

### super() — 부모 클래스의 메서드 호출
- super()는 부모 클래스의 메서드를 호출할 수 있게 해주는 것

```python
[작성법]
super().기반클래스의 메서드()
```

```python
# 사진 설명
```

```python
class Person:  # 부모
    def __init__(self, name, age):  # 이름·나이 초기화 , # 매개변수1, 매개변수2
        self.n = name
        self.a = age
    def greet(self):
        print(f'Hello, My name is {self.n}')
class Student(Person):  # 자식
    def __init__(self, name, age, major):  # 생성자 재정의(전공 추가) # 매개변수 1,2,3
        super().__init__(name, age)  # 이름·나이 저장은 부모 생성자에게 맡김 , # 인수1, 인수2
        self.m = major               # 자식만의 속성
    def study(self):                 # 자식 메서드
        print(f'{self.n}학생은 {self.m}를 공부합니다.')

        
s = Student('Alice', 20, 'Math')  # 인수3개 init()에 전달
s.greet()   # 부모 메서드
s.study()   # 자식 메서드
```

```python
<매개변수 vs 인수 헷갈 포인트>

- 매개변수 (Parameter): 함수나 메서드를 정의할 때 
  외부에서 값을 받기 위해 선언하는 변수

def __init__(self, name, age, major): 여기서 쓰인 name, age, major가 매개변수

부모 클래스의 def __init__(self, name, age):에 있는 name과 age도 매개변수

- 인수 (Argument): 함수나 메서드를 호출(실행)할 때 실제로 전달하는 값이나 변수

super().__init__(name, age)는 부모 클래스(Person)의 __init__ 메서드를 
'호출'하는 코드
-> 따라서 이때 괄호 안에 전달하는 name과 age는 인수
-> 마찬가지로 맨 아래 Student('Alice', 20, 'Math')에서 쓰인 
   'Alice', 20, 'Math'도 인수

즉, Student 객체를 생성할 때 인수로 받은 값을 Student의 매개변수로 받은 뒤, 
이를 다시 부모 클래스의 __init__ 메서드를 호출하면서 인수로 넘겨주고 있는 과정

⭐ 용어 정리: 인수와 매개변수

- 전달하는 쪽 = 인수(argument) / 받는 쪽 = 매개변수(parameter)
 super().__init__(name, age)의 name, age = 인수 2개를 전달하겠다 /
 부모 __init__(self, n, a)의 n, a = 매개변수로 받겠다
 ⇒ 둘의 이름이 같지 않아도 됨
- 매개변수는 값을 받아서 사용하는 것이고 변수이기 때문에, 변수 사용되는 곳에서는 이름을 명확하게 바꿔서 써야함→ subject를 s로 바꿨다면 안에서 쓰는 곳도 s로
```

```python
class Person:
    def greeting(self):
        print('안녕하세요')
class Student(Person):
    def greeting(self):
        super().greeting()       # 상위 클래스가 가지고 있는 메서드명의 내용을 가져다 쓰는 것
        print('저는 학생입니다.') # 자식 문장 추가
        
alice = Student()
alice.greeting() 

# 공통 기능은 부모에서 재사용, 다른 부분만 자식에서 추가 → 오버라이딩을 쓰는 이유
```

```python
### 연습문제1
- 클래스 C와 클래스 Car를 생성
- 클래스 Car는 클래스 C를 상속받음
- 클래스 C에는 생성자를 통해 데이터를 저장하는 기능 생성 => __init__
- 클래스 Car 는 info_print 메서드를 생성하고 출력기능을 만든다
- 출력되는 내용: 바퀴수 4와 가격 1000원
```

```python
[출력내용]
바퀴수 4
가격 1000
```

```python
class C:
    def __init__(self, w, p): # w,p이 매개변수 # self에 들어가는 건 현재 c1의 주소값
        self.wheel = w        # self.속성명 = 매개변수
        self.price = p
class Car(C):
    def info_print(self):
        print(f'바퀴수 : {self.wheel}')
        print(f'가격: {self.price}')
c1 = Car(4, 1000)   # 부모 C의 생성자가 자동 실행 -> 인수 전달 필수
c1.info_print()
```

```python
## 연습문제2
- 상속의 개념을 활용하여, 기본 인적 사항(이름,나이)을 가진 Person 클래스를 만들고,
  이를 상속받는 Student 클래스를 작성하여 학생 번호까지 출력하는 기능을 구현하세요
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self, name, age, s_id): # 생성자 재정의(값 3개)
        super().__init__(name, age)      # 부모 생성자 호출 , # 받은 name,age 인수값을 위의 Person class의 n,a 매개변수에 전달, Person은 받은 매개변수 n,a 값을 속성명인 self.name, self.age 에 저장
        self.s_id = s_id
    def print_info(self):
        print(f'이름: {self.name}')
        print(f'나이: {self.age}')
        print(f'번호: {self.s_id}')
s1 = Student('지수', 20, 332002)
s1.print_info()
        
```
