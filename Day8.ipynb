```python
[작성법]
class 클래스이름:                # 설계도
    def 메서드이름(self, 매개변수1, 매개변수2): # self 다음에 받을 값의 매개변수
        self.속성명1 = 매개변수1 # 해당 매개변수1을 객체(self)의 속성명1에 저장한다 , # 객체 안의 저장 공간에 저장
        self.속성명2 = 매개변수2 # 두 번째 속성에 저장
객체명 = 클래스이름()            # 객체 생성
객체명.메서드이름(인수1, 인수2)  # 객체명은 self에 들어가고, 인수1은 매개변수1, 인수2은 매개변수2에 매칭
```

```python
- 위 메서드 = 값을 받아서 저장하는 역할 → 어디에? self.속성명이라는 속성
- self에는 객체의 주소값이 들어감
```

```python
class Person:
    def setdata(self, name, age): # 메서드1 -> 저장 기능: 이름·나이를 받음
        self.name = name   # 객체(self)의 속성 name ← 매개변수 name
        self.age = age     # 객체의 속성 age ← 매개변수 age
    def print_function(self): # 메서드2 -> 출력 기능: 저장된 값을 사용하므로 self만
        print(f'이름은 {self.name}이고, 나이는 {self.age}살입니다.') 
        # 위의 setdata 메서드가 매개변수에 값 저장 했으므로 또 name과 age를 정의할 필요가 없다. 
        # self.속성명으로 이 객체의 값을 꺼냄
        
#객체1 생성
p1 = Person()           # Person 설계도로 객체 p1 생성
p1.setdata("Alice", 25) # self=p1, name='Alice', age=25 → p1 안에 저장
p1.print_function()     # 객체명. 만 앞에 붙여주면 함수랑 똑같다
```

```python
# 설계도 생성
class Person:
    def setdata(self, name, age, job): #저장기능 # 받을 값 3개 → 매개변수 3개 (+ self) (class에서만,함수는 걍 공백)
        self.name = name # 객체명.속성명 = 매개변수 (# 왼쪽 = 객체의 속성, 오른쪽 = 매개변수), 속성(= 변수)
        self.age = age 
        self.job = job
    def print_function(self): #출력기능, 위에서 이미 매개변수에 인수를 저장했으므로 가져다쓰면 되기 때문에 따로 또 매개변수 쓰지 않아도 됌
        print(f'이름은 {self.name}이고, 나이는 {self.age}입니다.') # 누구의 속성인지 명확하게 self.속성명으로 적어야함
    def print_job(self): #출력기능
        print(f'직업은 {self.job}입니다.') 

# 객체 생성
p3 = Person()
p3.setdata("James", 28, 'programmer') # self <- p3 , name <- James, age <- 28, job <- programmer 순서대로  매칭
p3.print_function()  # 이름·나이 출력
p3.print_job()       # 직업 출력 (메서드이므로 괄호 잊지 않기)
```

```python
print(p3.name) # 객체의 속성명에 값이 들어가 있으므로 바로 출력도 가능
print(p3.age)
print(p3.job)
```

```python
balance1 = 1000 # 계좌 1의 잔고 (전역변수)
def deposit1(amount):  # 입금 함수 — 입금액 amount를 받음
    global balance1    # 함수 안에서 전역변수를 수정하겠다고 선언 (Day6 global)
    balance1 += amount # 잔고에 입금액 더하기
    return balance1    # 바뀐 잔고 반환
def withdraw1(amount): # 출금 함수
    global balance1    # 전역 잔고 사용
    if balance1 >= amount:  # 출금 가능한지 잔고와 비교
        balance1 -= amount  # 가능하면 차감
        return balance1     # 남은 잔고 반환
print(deposit1(500))        # 1000 + 500 → 1500
```

```python
balance2 = 10000      # 계좌 2의 잔고 — 변수를 새로 만들어야 함
def deposit2(amount): #입금 함수
    global balance2   # 전역 balance2 사용
    balance2 += amount  # 입금
    return balance2     # 잔고 반환
def withdraw2(amount): #출금
    global balance2    # 전역 balance2 사용
    if balance2 >= amount: # 잔고 확인
        balance2 -= amount # 차감
        return balance2    # 남은 잔고 반환
print(deposit2(5000))      # 10000 + 5000 → 15000
```

```python
02. 클래스를 사용하여 계좌 구현
```

```python
# 설계도 생성
class Account: # 계좌 설계도
    def setdata(self, owner, balance = 0):  # 예금주·잔고 저장, 처음 입금 안 하는 경우: 잔고는 디폴트 0(디폴트 매개변수 -> 기본값 설정) ,  # self <- acc1 매칭, owner <- 'Alice',   balance <- 800 
        self.owner = owner      # 이 객체의 예금주
        self.balance = balance  # 이 객체의 잔고
    def deposit(self, amount):  # 입금 메서드 , # self <- acc1 , 밑에 deposit 기능의 인수 500 받음
        self.balance += amount  # 객체의 잔고에 더해 저장 , # amount는 지역변수: amount 는 한번 쓰고 사라짐(종료) 
        return self.balance     # 총 1300 그 값을 return, 반환값 -> acc1.deposit(500)에 전달 -> # 바뀐 잔고를 호출한 곳으로
    def withdraw(self, amount): # 출금 메서드
        if self.balance >= amount: # 이 객체의 잔고로 출금 가능한지
            self.balance -= amount # 차감해서 객체에 저장
        else:
            print('잔고가 부족합니다.')
        return self.balance # (if 밖) 출금 여부와 관계없이 현재 잔고 반환
        
# 객체 생성
acc1 = Account()            # 통장 1 찍어 내기
acc1.setdata('Alice', 800)  # 예금주 Alice, 잔고 500으로 초기 세팅 , #저장 메서드 사용, 저장할 값의 인수 전달
amount1 = acc1.deposit(500) # 우변에 return값이 들어있음, 오른쪽에 들고있는 값을 좌변에 저장 -> # 500 입금 → 반환된 1300 저장
print(amount1)
```

```python
## 03 ㅣ 사칙연산 계산기 만들기
### 01. 계산기 설계
- 사칙연산이 가능
- 속성 2개: 두개의 값을 저장하기
- 메서드 4개: 덧셈, 뺄셈, 곱셈, 나눗셈
```

```python
class FourCal(): # 계산기 설계도 (빈 괄호는 있어도 없어도 됨)
    def setdata(self, first, second): # 두 값을 받아 저장하는 기능
        self.first = first            # 첫 번째 값 → 객체의 first
        self.second = second          # 두 번째 값 → 객체의 second
    def add(self):             # 덧셈 기능 (저장된 값을 쓰므로 self만)
        self.result = self.first+ self.second # 결과를 객체의 result 속성에 저장 , # 객체에 저장할라면 self.result라 써야함 (어디에 저장할껀지 명시해야함)
        return self.result     
    def mul(self):
        self.result = self.first* self.second # 곱을 result에 저장 (덮어씀)
        return self.result
    def sub(self):
        self.result = self.first-self.second
        return self.result
    def div(self):
        self.result = self.first/self.second
        return self.result

# 입력받아서 전달하기
num1 = int(input('첫번째 값을 입력하세요'))
num2 = int(input('두번째 값을 입력하세요'))

cal = FourCal()
cal.setdata(num1, num2) # 입력값을 매개변수로 전달,  # first=num1, second=num2 저장
print(cal.add()) #cal.add() 에 add 메서드의 self.result 반환값 
print(cal.mul())
print(cal.sub())
print(cal.div())
```

```python
## 연습문제 ㅣ
- CPU, RAM, SSD를 구성요소로 가지고 있는 Computer 클래스를 이용해서 인스턴스를 생성
- Computer  클래스는 각 구성요소의 값을 저장할 수 있는 set_spec() 메서드와
- 구성요소의 값을 출력할 수 있는 hardware_info() 메서드로 구성 
[출력결과]
CPU = i7
RAM = 16GB
SSD = 512GB
```

```python
class Computer:
    def set_spec(self,CPU,RAM,SSD):
        self.CPU = CPU
        self.RAM = RAM
        self.SSD = SSD
    def hardware_info(self):
        print('CPU = ', self.CPU) # or print(f'CPU= {self.CPU}')
        print('RAM = ', self.RAM)
        print('SSD = ',self.SSD)
m = Computer()
m.set_spec('i7', '16GB', '512GB')
m.hardware_info()
```

```python
## 연습문제 | Student 클래스
- Student 클래스를 만들어서 name, age,score라는 속성을 가짐
- setdata() 메서드를 정의하여 학생의 이름, 나이, 점수를 설정
- is_passed() 메서드를 추가하여 점수가 60 이상이면 Passed를, 그렇지 않으면 Failed 를 출력하는 기능을 구현
  ('John', 20, 75) / ('Alice', 22, 55)
  [출력결과]
  John Passed
  Alice Failed
```

```python
class Student:
    def setdata(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

    def is_passed(self):
        if self.score >= 60:
            print(f"{self.name} Passed")
        else:
            print(f"{self.name} Failed")


s1 = Student()
s1.setdata('John', 20, 75)
s1.is_passed()

s2 = Student()
s2.setdata('Alice', 22, 55)
s2.is_passed()
```

```python
학생이 여러 명이면 반복문으로 처리할 수도 있음.
```

```python
students = [('John', 20, 75), ('Alice', 22, 55)...]

for name, age, score in students:
    s = Student()
    s.setdata(name, age, score)
    s.is_passed()
```
