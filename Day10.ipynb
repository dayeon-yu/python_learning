```python
1.
class Contact 클래스 생성
  __init__ 속성: name, phone
```

```python
2.
- 연락처 3개 빈리스트(contacts)에 저장 저장
for문 사용
    name, phone입력받기
    빈리스트에추가(객체생성)
```

```python
class Contact: 
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
contacts = []
#연락처 3개 저장
for i in range(3):
    name = input("이름: ")
    phone = input("전화번호: ")
    #객체생성후 추가
    #s = Contact(name, phone) # 객체명 = 클래스로 인해 만들어진 객체
    contacts.append(Contact(name, phone))
```

```python
print(contacts)
```

```python
for i in contacts:
    print(i.phone)
```

```python
### 문제03 | 최고 점수 학생 찾기
- class Student 생성
  속성: name, score
- 학생 3명의 이름, 점수 입력받기
- 빈리스트(students)에 저장
- 최고 점수 찾기위한 값 비교
- 최고 점수 학생 출력

이름: 수지
점수: 95
이름: 지아
점수: 88
이름: 유나
점수: 100
최고 점수 학생: 유나 (100점)
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
```

```python
students = []# [객체1, 객체2, 객체3]
for i in range(3):
    name = input("이름: ")
    score = int(input("점수: "))
    students.append(Student(name,score))
```

```python
#최고점수 찾기 
top = students[0]
for s in students[1:]: #[지아객체, 유나객체]
    if s.score > top.score:
        top = s
print(f"최고점수 학생 : {top.name} ({top.score}점)")
```
