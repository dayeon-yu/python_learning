## 01 | OpenAPI란?

```python
01)API란?
- Application Programming Interface의 약자 
- 서로 다른 프로그램(소프트웨어)끼리 데이터를 주고 받거나 기능을 사용할 수 있도록
정해놓은 약속(규칙)
- 프로그램과 프로그램 사이를 이어주는 중간다리 역할 

비유
식당에서 손님이 직접 주방에 들어가지 않고, 웨이터에게 주문하면 웨이터가 주방과 손님
사이를 중개해줌 
-> 웨이터 = API -> 메뉴판으로 요청 -> 음식받음
                  정해진 방식으로 요청 -> 정해진 방식의 결과를 받음
```

```python
02) OPEN API 
- 누구나 사용할 수 있도록 공개된 API
- 사용자가 서버로부터 표준화된 데이터를 프로그래밍적으로 받아 볼 수 있음
- 개방된 오픈 API를 이용해 다양한 서비스, 애플리케이션, 다양한 형태의 플랫폼을 개발 
할 수 있음 
```

```python
information_ls = [] #병원 정보를 담을 빈 리스트
for i in hospital:
    information_dict = {} #각 병원에 대해 새로운 딕셔너리 
    information_dict["기관명"] = i["DUTYNAME"] #기관명 : 병원의 이름
    information_dict["병원분류명"] = i["DUTYDIVNAM"] #병원분류명
    information_dict["간이약도"] = i["DUTYMAPIMG"]
    information_dict["주소"] = i["DUTYADDR"]
    #전화번호
    information_dict["전화번호"] = i["DUTYTEL1"]
    #
    information_ls.append(information_dict)
information_ls
```
