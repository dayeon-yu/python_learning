## 02 ㅣ 문자열 메서드

```python
- count(): 특정 문자열이 등장하는 횟수 반환
- find(): 문자열에서 처음 등장하는 위치 반환(없으면 -1 반환)
- index(): find와 유사, 없으면 valueError를 발생
- join(): 문자열 사이에 값을 넣어 반복가능객체의 요소들을 연결
- split(): 문자열 내 특정 구분자를 기준으로 문자열 나눔
- replace(): 문자열을 다른 문자열로 교체
- 불필요한 문자를 제거: lstrip(), rstrip(), strip()
- 대소문자 변환메서드: lower(), upper(), capitalize()
- 기타
```

```python
count(): 특정 요소의 개수를 세는데 사용

[작성법]
결과값을 담을 변수명 = 문자열변수.count("문자,문자열")
# 한 글자든 단어든 몇 번 나오는지 정수로 반환
```

```python
text = "hello world, hello python"
# 'hello'가 2번 들어 있는 문자열
count_hello = text.count('hello')
# 단어 'hello'의 등장 횟수 2를 반환 → 저장
print("number of 'hello': ", count_hello)
```

```python
text = "hello world, hello python"
count_o = text.count('o')
# 글자 'o' 개수: hello world, hello python → 4 반환받아 저장
print("number of 'o': ", count_o)
```

```python
find(): 특정 문자열을 찾고자 할 때 사용

[작성법]
결과값변수 = 문자열변수명.find("찾을 글자(단어)")
# 처음 나오는 인덱스 반환
결과값변수 = 문자열변수명.find("찾을 글자(단어)", 기준 인덱스) 
#기준 인덱스 이후부터 찾아라, # 두 번째 값 = 검색 시작 위치

* 찾는 문자열이 없는 경우 -1 반환 
# 오류가 나지 않음 (index와 차이)
```

```python
## 문자 찾기
text = "hello world" 
index = text.find('l')  #맨 처음나오는 인덱스 반환 후 정지
print(index)
```

```python
join(): 여러 문자열을 하나의 문자열로 결합할 때 사용

[작성법]
결과값변수 = "구분자".join(변경하고 싶은 변수명)
# 구분자 문자열이 앞, 합칠 대상(리스트 등)은 괄호 안
```

```python
<사용가능한 구분자>
- "": 쌍따옴표
- " ": 공백
-',': 쉼표
-'/n': 새줄문자
```

```python
# 리스트의 문자열 요소들을 공백으로 결합
w = ["hello", "world", "python", "is", "awesome"]
sentence = ' '.join(w) 
# 요소 사이사이에 공백 1칸을 넣어 하나의 문자열로
print(sentence)
```

```python
w2 = ['h','e', 'l', 'l', 'o'] # 글자 하나씩 든 리스트
sen2 = ''.join(w2) 
# 빈 문자열 구분자 → 사이에 아무것도 없이 붙임
print(sen2)
```

```python
w = ["hello", "world", "python", "is", "awesome"]
# 쉼표사용
a = ','.join(w)
print(a)
```

```python
str1 = 'hello my name...' 
# 문자열도 반복 가능 객체 → 글자 하나하나가 요소
str2 = '\n'.join(str1)    
# 글자 사이마다 줄바꿈 삽입
print(str2)               
# 한 줄에 한 글자씩 (공백도 한 글자 → 빈 줄처럼 보임)
```

```python
split(): 문자열 분리

[작성법]
변수명 = 문자열변수.split() -> 공백을 기준으로 잘라줘 
# 괄호 비우면 공백(스페이스·탭·줄바꿈) 기준
변수명 = 문자열변수.split('구분자') 
# 지정한 구분자 기준
변수명 = 문자열변수.split('구분자', maxsplit = 숫자값)
# 최대 분리 횟수 제한
```

```python
## 공백 기준으로 분리

text = "hello world, hello python"
word = text.split() # 공백 기준으로 잘라 리스트로 반환
print(word)
```

```python
##특정 문자를 기준으로 문자열 분리

str1 = "사과-딸기-수박-오렌지-키위"
arr1 = str1.split('-') # '-' 기준으로 분리 -> 리스트로 반환
print(arr1)
```

```python
## 분리개수 제한-> 최대 2번 분리

text = "one, two, three, four, five"
split_text = text.split(', ', maxsplit = 2)
# ', '(쉼표+공백) 기준, 2번만 자름 → 조각 3개
print(split_text)
```

```python
print(split_text[2]) 
# 실제로 리스트의 세번째는 maxsplit 2회 하고 남은 나머지 전체가 다 묶인것
# 2번 인덱스 = 남은 전체
```

```python
보충: split ↔ join은 반대 관계

"a-b-c".split('-') → ['a', 'b', 'c'] 
'-'.join(['a', 'b', 'c']) → "a-b-c"


input().split() → 한 줄에 여러 값을 입력받을 때 자주 사용 
(예: a, b = input().split() → 언패킹)
```

```python
a,b = input('a, b값을 입력하세요: ').split() 
# input으로 값을 받음->  공백 기준으로 split->  리스트로 반환 -> 언패킹
print(a, b)
```

## 03 ㅣ 함수

```python
- 특정 작업을 수행하는 코드 블록을 의미
- 프로그래밍에서 함수는 코드의 재사용성을 높임
- 프로그램 더 이해하기 쉽고 관리하기 쉬움
-> 함수를 사용 동일한 코드를 반복적으로 작성하지 않고도 필요할 때마다 해당 기능을 호출하여 사용 가능
```

```python
01) 함수 용어 정리
- 내장함수: 이미 만들어져 있어서 사용법대로 사용하면 되는 함수 (print, len, range …)
- 사용자 함수: 사용자가 직접 만든 함수
- 함수 정의: 사용자 함수를 새롭게 만드는 것 (def)
- 인수: 사용자 함수에 전달할 입력(argument)  — 호출할 때 괄호에 넣는 값
- 매개변수: 인수를 받아서 저장하는 변수 (parameter) — 정의할 때 괄호에 쓰는 변수
- 반환값: 사용자 함수의 출력(리턴값)
- 함수 호출: 만들어진 사용자 함수를 실제로 사용하는 것
```

```python
 02) 함수를 사용하는 이유
- 중복코드 제거: 동일한 기능을 여러번 작성할 필요 없이, 한번 정의한 함수를 여러곳에서 호출해서 사용 가능
- 디버깅 및 유지보수 용이: 함수 안에 문제가 생기면 그 부분만 수정하면 되므로, 디버깅, 유지보수가 쉬워짐
* 디버깅: 오류(버그)를 찾고 원인을 분석하여 수정하는 과정
```

```python
03) 함수정의 하기
- def 는 함수를 만들 때 사용하는 예약어 (define의 약자)
- 함수 이름은 함수를 만드는 사람이 임의로 만들 수 있음
```

```python
[작성법]
def 함수이름(매개변수1):
    수행할 문장1
    수행할 문장2
    ...
    return 반환값        # 결과를 호출한 곳으로 돌려주고 함수 종료

#호출
변수 = 함수이름(인수)
```

```python
04) 함수 호출하기
- 의미: 정의된 함수를 사용한다는 것!

[작성법]
#1. 반환값이 있는 경우 
변수 = 함수이름(인수)      # 반환값 담는 변수 바구니     
                      # 돌려받은 값을 변수에 저장 (또는 print로 바로 출력)

#2. 반환값이 없는 경우 
함수이름(인수)             # 돌려줄 값 없으니 변수 바구니 필요 x

#3. 매개변수 없는 경우
변수 = 함수이름()          # 인수 필요 x
```

```python
def greet(greeting, *args):   # 일반 매개변수 먼저, *args는 마지막
    print(greeting)           # 첫 번째 인수
    print(args)               # 나머지 인수들의 튜플
    for name in args:         # 튜플에서 이름을 하나씩 꺼내
        print(f'{greeting}, {name}!')
greet('hello', 'Jane','Alice','James') 
# 첫번째 값만 GREETING에 할당, 나머지 값들 모두를 *ARGS에 할당
# 'hello' → greeting, 나머지 3개 → args
```

## 04 ㅣ 지역변수와 전역변수

```python
- 변수의 종류에 따라 사용범위와 라이프사이클이 달라짐
 1. 지역변수 (Local Variable)
- 함수 내부에서 선언된 변수, 함수가 실행될 때 생성되고, 함수가 종료되면 사멸
 2. 전역변수(Global Variable)
- 함수 외부에서 선언된 변수, 프로그램 전체에서 접근할 수 있음
- 프로그램이 종료될 때까지 메모리에 존재
```

```python
var = 'global' # 전역변수
def function():
    var = 'local' 
    # 지역변수-> 함수 호출해야만 볼 수 있고 함수 종료되면 메모리에서 사라짐
    print(var)
function()
print(var)
```

```python
01) 지역변수
- 정의: 함수나 블록 내부에서 선언된 변수
- 범위: 변수가 선언된 함수나 블록 내부에서만 접근 가능
- 라이프 사이클: 함수나 블록이 실행될 때 생성되고, 종료되면 소멸
```

```python
def my_function(): # 함수 정의
    local_v = 10   # 지역변수 (함수 안에서만 존재)
    print(f"지역변수: {local_v}")
my_function()      # 호출 → '지역변수: 10' 출력 후 local_v 소멸
print(local_v)     # ❌ 함수 밖에서 지역변수 사용 → 오류
```

```python
02) 전역변수
정의: 함수나 클래스 외부에서 선언된 변수
범위: 프로그램 전체에서 접근 가능
라이프 사이클: 프로그램이 시작할 때 생성되고, 프로그램이 종료될 때까지 유지.
```

```python
## 연습문제 02 | 모든 입력의 평균값 구하기
- 주어진 모든 수의 평균값을 계산해 주는 함수를 작성
- 주어진 수의 개수는 정해져 있지 않다
# TEST코드
```

```python
def calculate_average(*args):  # 인수 개수 제한 없음 → 튜플 args로 받음
    if not args:    #  입력이 없으면 args == False, not args == True
                # 빈 튜플이면 (Day1: 비어 있는 값 = False → not False = True)
        return None
    total = 0 
    count = 0
    for number in args:
        total += number
        count += 1
    avg = total / count
    return avg
print(calculate_average(1, 2, 3, 4, 5))  # 출력: 3.0
print(calculate_average(10, 20, 30))    # 출력: 20.0
print(calculate_average())              # 출력: None (입력이 없는 경우)
```

```python
### 연습문제 3 | 문자열 뒤집기 
주어진 문자열을 뒤집은 후 반환하는 함수를 작성하세요
-  [::-1] 사용 금지
- 반복문을 사용하여 구현하세요.
```

```python
# 방법 1
def reverse_text(text):      # 'python' 문자열을 받음
    result = ""              # 결과를 모을 빈 문자열
    for ch in text:         # 한 글자씩 꺼내 (p, y, t, h, o, n)
        result = ch+result  # 꺼낸 글자를 기존 결과의 앞에 붙임 → 순서가 뒤집힘
        print(result)       # 중간 과정 확인용 출력 (p → yp → typ …)
    
    return result           # 반환

print(reverse_text("python"))   # nohtyp
print(reverse_text("apple"))    # elppa
```

```python
# 방법 2
def reverse_text(text):  # 'python' 문자열을 받음
    a = list(text)       # 글자 리스트 ['p','y','t','h','o','n']
    result = []          # 결과 리스트
    for i in a[:]:     # 복제본(a[:])으로 6번 반복 (a는 pop으로 줄어들기 때문)
        b = a.pop()   # 맨 뒤 글자를 꺼냄 (n → o → h …) — pop은 꺼낸 값을 반환
        result.append(b)  # 꺼낸 순서대로 추가 → 뒤집힌 순서
    return ''.join(result) # 리스트를 빈 문자열 구분자로 합쳐 문자열로 반환

print(reverse_text("python"))   # nohtyp
print(reverse_text("apple"))    # elppa
```
