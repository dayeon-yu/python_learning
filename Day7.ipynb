```python
01) 리턴의 특징
- 전달되는 인수가 없어도 함수를 호출하면 리턴값이 반환되도록 만들 수 있음
```

```python
def something():  # 매개변수 없는 함수
    return 2
r = something()  # 반환값을 담을 변수(바구니) 준비 → r에 2 저장
print(r)
```

```python
02) 리턴의 사용
1. 리턴값이 2개인 경우
```

```python
def add_sub(a,b):
    c = a+b
    d = a-b
    return c,d  #쉼표를 사용해 나열하면 값이 튜플로 묶임 ->(c, d) 튜플로 묶여 반환
x,y = add_sub(10,20) # 쪼개서 쓰고 싶으면 언패킹
print(x)
print(y)
```

```python
2. 리턴은 한번만 동작
-> 리턴을 만나면 '여기까지 동작하고 함수를 종료 시켜' 라는 의미
```

```python
def add_sub (a,b):
    return a+b      # 여기서 함수 끝 -> return은 이어서 출력이 안된다는 점 기억!! 
    return a-b      # ⚠️ 절대 실행되지 않는 줄
y = add_sub(10,20)  # 첫 번째 return 값만 받음
print(y)
```

```python
- return을 만나는 순간 함수는 더 이상 동작하지 못함
  그래서 돌려준 값은 30 하나뿐
- return을 두 줄 이어 써서 두 값을 다 전달하는 것은 불가능
  → 두 개를 보내려면 위의 return c, d(튜플) 방식
```

## 02 ㅣ 재귀호출

```python
01) 재귀 호출이란?
- 자기 자신을 호출하는 함수
- 반복적 / 규칙적 문제를 작은 하위 문제로 나누어 해결할 때 사용가능
```

```python
02) 특징
- 큰문제 -> 작은문제 -> 더작은문제 -> 종료조건
- 종료 조건이 반드시 필요!! 
- 종료 조건이 없으면 무한 호출
- 스택 구조로 동작 -> 함수 호출이 쌓이고 종료 조건에 도달하면 하나씩 반환
```

```python
-지금까지는 함수를 바깥에서 불렀지만, 재귀는 함수 안에서 자기 이름을 다시 부름
-문제가 점점 작아지다가 언젠가는 멈춰야 함 → 종료 조건을 만나면 재귀가 멈춤
-함수가 호출될 때마다 스택에 쌓이고, 
 종료 조건에 도달하면 하나씩 반환되며 마지막에 종료
```

```python
04) 재귀호출 사용하기
1부터 n까지 출력하기
5 -> 4 -> 3 -> 2 -> 1
0이 되면 종료
```

```python
def recursive(n):   # 숫자 n을 받음
    if n ==0:       # 종료 조건: 0이 되면 return 으로 종료하겠다
        return 0    # 값을 돌려주지만 받을 변수가 없어 사용 못 함
    else:
        recursive(n-1) # 먼저 끝까지 파고듦 # 5->4->3->2->1로 인수를 줄여서 전달(함수 재귀호출(반복) -> 스택에 정보 저장)
        print(n)       # 되돌아오면서 출력 → 그래서 1부터 나옴
recursive(5)
```

```python
⭐ 사람에게 넘기는 비유

- 첫 번째 사람이 5를 들고 있다가 옆 사람에게 4를 주며 "이거 가지고 일해" 라고 넘김
  4를 받은 사람은 다음 사람에게 3을, 그 다음은 2, 그 다음은 1을 넘김
- 1을 가진 사람이 다음에 0을 넘기면 → 종료 조건을 만나 멈춤
- 그리고 거꾸로 돌아오면서 각자 자기 값을 출력 → 1, 2, 3, 4, 5 순서
```

```python
# 재귀의 핵심 원리
: 함수가 호출되면 스택에 정보가 쌓이고 함수가 종료되면 스택에 정보가 사라짐
```

```python
⭐정리 — 되돌아오는 '위치'가 핵심 (얘만 보고 이해되면 밑에 안봐도 ok)

1) 함수가 호출되면 지역변수(n)가 스택에 쌓임 → 5, 4, 3, 2, 1, 0 순서로 쌓임
2) n == 0에서 True → return 0 → 값을 돌려주지만 받을 바구니가 없어 사용하지 못함
3) 그 프레임이 사라지고, 돌아오는 위치는 recursive(n-1)을 쓴 그 줄
4) 그 아래에 print(n)이 남아 있으므로 그때 출력됨 → 현재 프레임의 n값(1)이 출력
5) 출력하고 나면 더 수행할 코드가 없어 그 프레임도 스택에서 제거 
   → 같은 과정이 5까지 반복
```

```python
# 위의 재귀 호출 개념 설명 (머릿속에 그림 떠올리기)
일단 함수 안에 자기 자신을 또 호출하여 다시 자기자신으로 올라가 기능을 수행하는 것을 재귀 호출이라 함
쉽게 말해, 아래의 코드가 호출될때마다 블럭처럼 여러개 생성되어 호출하면 다음 단계로 이어져 나간다 생각
def recursive(n):
    if n ==0:
        return 0 
    else:
        recursive(n-1)  # 호출되는 만큼 이 블럭이 계속 반복 
        # -> 이 부분 만나면 일단 다시 함수가 기능하므로
        # print(n)은 맨 나중에 함수 종료되면 다시 거꾸로 되돌아가며 실행된다
        print(n)
```

```python
인수로 받은 5->4->3->2->1 은 n= 5부터 순서대로 stack에 쌓임
함수가 종료되면 스택에 정보가 사라짐
-> 마지막 함수 블록이 종료되면 스택 맨 위의 n=0 이 사라짐, 
-> 이후 함수 블록이 0->1->2->3->4->5 순서대로 다시 회귀
-> 회귀하는데 자기 전 블록의 recursive(n-1)부분으로 돌아가 그 후에 1->2->3->4->5 순으로 print 출력
-> 블록마다 n값 출력하면 함수가 종료되므로 한 함수 블록씩 종료되며 stack 에서도 위에부터 값이 삭제
```

```python
# 위의 코드 설명 
1) 인수 5 받고 5->4->3->2->1로 인수를 줄여서 전달(재귀함수 반복될 때 마다 위의 함수 코드 그대로 여러개 생겨난다 생각)
2) n = 0 이면 return 만나 재귀함수 종료 -> 반환값 0을 그 전 단계에 반환 (but 바구니(변수)가 없으므로 받아서 쓸 수는 없다.)
3) n = 1부터 다시 원래 있던 위치인 recursive(n-1) 부분으로 돌아옴-> 이후 print(n)만나서 출력 후 해당 단계 함수 종료 , 
4) 계속 전 단계로 돌아가며 1,2,3,4,5 차례로 출력되고 한번씩 출력될때마다 함수 종료 (한단계씩 함수 종료될 때 스택에 쌓인 정보도 함께 삭제)
```

```python
<recursive(5) 실행 중 스택의 변화>
[쌓이는 중 — 아직 아무것도 출력되지 않음]
# 안쪽으로 파고드는 단계
 n=0  <- 종료 조건 만남! return 0
# 맨 위, 여기서 멈춤
 n=1
# recursive(0)을 부른 상태로 대기
 n=2
# recursive(1)을 부른 상태로 대기
 n=3
# 〃
 n=4
# 〃
 n=5  <- 여기부터 시작:  처음 호출
# 맨 아래
 
[되돌아오는 중 — 이때 print 실행]
# 되돌아오며 출력
 n=0 프레임 제거 -> n=1 자리로 복귀 -> print(1)
# 1 출력 후 n=1 프레임 제거
 -> n=2 자리로 복귀 -> print(2)
# 2 출력
 -> n=3 자리로 복귀 -> print(3)
# 3 출력
 -> n=4 자리로 복귀 -> print(4)
# 4 출력
 -> n=5 자리로 복귀 -> print(5)
```

```python
# 바구니 만들기

def recursive(n): # 인수를 받음
    if n ==0:     # 종료 조건
        return 0 
    else:
        c = recursive(n-1)  # 반환받는 자리 (우변이 반환값 들고있음) , 반환값 돌려받아 바구니(C)에 저장
        print(n)
        return c  # 바구니 있을시 -> return 있어야함!! 돌려줘야함
a = recursive(5)  # 최종 반환값을 a에 저장
print(a)
```

```python
⭐ 바구니와 return은 한 쌍

- 앞의 코드는 돌려받을 자리(변수)가 없어서 반환값을 쓸 수 없었음
- 값을 쓰려면 호출한 자리에 바구니(c =) 를 두어야 함 → 돌아오는 위치가 바로 그 줄
- 그리고 return c까지 써야 그 값이 한 단계씩 밖으로 전달됨 → 없으면 중간에서 끊김
- 재귀이므로 모든 단계에 똑같이 바구니와 return이 하나씩 준비되어 있는 셈
- 그래서 0이 0 → 1 → 2 → 3 → 4 → 5 단계를 거쳐 맨 밖의 a까지 도착
```

```python
재귀호출 -> 들어가서 동작하고 동작 후에는 다시 돌아나오기 
(무한루프에 빠지지 않도록 종료조건 명시)
```

```python
- recursive(n-1) 이 끝나고 나면 그 자리에 반환값이 들려 있는 것 
  → 그걸 바구니에 담을 뿐
- 반환값이 있으면 print(함수())처럼 바로 출력도 가능 
  → 그동안 써 온 return과 완전히 같은 원리
```

```python
def factorial(n):   # n을 받아 팩토리얼 계산
    #1. 종료조건
    if n == 1:
        return 1


    #재귀호출
    else:
        return n * factorial(n-1) 
        # 현재 n × (한 단계 작은 결과) — 계산과 반환을 한 줄로

result = factorial(4)
print(result)
# [결과] 24
        
```

```python
📘 보충: 재귀 팩토리얼과 반복문 팩토리얼


def factorial_loop(n):          # 반복문 방식
    result = 1                  # 곱셈의 시작값은 1
    for i in range(1, n+1):     # 1부터 n까지
        result *= i             # 누적해서 곱함
    return result               # 결과 반환
print(factorial_loop(4))


-> 재귀와 같은 결과
```

```python
## 추가 | enumerate(): 인덱스번호,값
- 리스트나 문자열 같은 반복 가능한 객체에서 인덱스 번호와 값을 동시에 추출
- 앞이 인덱스, 뒤가 값 → for i, v in enumerate(...) 형태로 반복문과 결합하여 사용
- (인덱스, 값) 순서로 튜플을 반환
- start를 주면 인덱스 시작 번호를 바꿀 수 있음(목록을 1번부터 보여 줄 때 유용)
  : enumerate(객체, start = 숫자값)과 같이 start 값을 지정하면 
    인덱스 시작 번호를 원하는 숫자로 변경할 수 있음
```

```python
list2 = [10,20,30]
for i, v in enumerate(list2):   # (인덱스, 값) 두 개를 함께 꺼냄
    print(f'인덱스:{i}, 밸류:{v}')
```

```python
list2 = [10,20,30]
for i, v in enumerate(list2, start = 1): # 시작 인덱스 번호 바꿔서 사용 가능
                                         # start=1 → 1부터
    print(f'인덱스:{i}, 밸류:{v}')
```

```python
# 문자열과 함께 사용하기

for i, ch in enumerate('abc'):    # 문자열도 반복 가능 객체
    print(f'인덱스:{i}, 문자:{ch}')
```

```python
# 문자열과 함께 사용하기

a = []                        # 결과를 담을 빈 리스트
for i, ch in enumerate('abcdeFGHI'): 
    print(f'인덱스:{i}, 문자:{ch}')
    if i%2 == 0:              # 인덱스가 짝수면
        a.append(ch.upper())  # 대문자로 바꿔 추가
    else:                     # 홀수면
        a.append(ch.lower())  # 소문자로 바꿔 추가

print(a)         # 리스트 형태로 확인
b = ''.join(a)   # join으로 다시 하나의 문자열로 합침
print(b)
```

```python
## 문제 07 | 홀수 vs 짝수
- 정수 리스트: num_list 일때, 
짝수 번째 원소들의 합과 홀수 번째 원소들의 합 중 큰값을 return 하도록 solution 함수 작성
```

```python
def solution(num_list):
    odd_sum = 0   # 홀수 저장공간
    even_sum = 0  # 짝수 저장공간
    for i, v in enumerate(num_list): # 인덱스와 밸류를 같이 꺼냄 -> (인덱스, 값)
        if i%2 ==0:
            even_sum += v

        else:
            odd_sum += v
    answer = max(odd_sum, even_sum)   # 둘 중 큰 값 선택
    return answer                     # 반환

# 테스트
print(solution([4,2,6,1,7,6])) # 17
print(solution([-1,2,5,6,3]))  # 8
```

```python
def solution(num_list):
    answer = 0
    odd_sum, even_sum = 0, 0
    for i in range(len(num_list)):
        if i % 2 == 0:   # 인덱스가 짝수 → 홀수 번째 원소
            even_sum += num_list[i]
        else:            # 인덱스가 홀수 → 짝수 번째 원소
            odd_sum += num_list[i]
    answer = max(odd_sum, even_sum)
    return answer
    
# 테스트
print(solution([4,2,6,1,7,6])) # 17
print(solution([-1,2,5,6,3]))  # 8
```

```python
📘 보충: 슬라이싱으로 더 짧게 쓰기

def solution(num_list):      # 정수 리스트를 받음
    return max(sum(num_list[0::2]), sum(num_list[1::2]))
    # 0번부터 2칸씩, 1번부터 2칸씩 각각 합해 큰 값
```
